### The latest rates

```python
import requests
import json

BASE = "https://api.frankfurter.dev/v2"

r = requests.get(f"{BASE}/rates", params={"providers": "ecb", "quotes": "idr,usd,sgd"}, timeout=30)
print(r.status_code)
print(r.url)
data = r.json()
print(type(data))
print(json.dumps(data, indent=2)[:1500])
```

### 1 month of daily rates

```python
r = requests.get(
    f"{BASE}/rates",
    params={"providers": "ecb", "from": "2026-09-01", "to": "2026-09-30", "quotes": "idr,usd,sgd"},
    timeout=30,
)
print(r.status_code, r.url)
data = r.json()
print(type(data), len(data))
sample = data[:3] if isinstance(data, list) else data
print(json.dumps(sample, indent=2)[:1500])
```

### Which currencies exist

```python
r = requests.get(f"{BASE}/currencies", timeout=30)
print(r.status_code)
data = r.json()
print(type(data), len(data))
sample = data[:3] if isinstance(data, list) else data
print(json.dumps(sample, indent=2)[:1500])
```

### Is a full year returned day by day?

```python
r = requests.get(
    f"{BASE}/rates",
    params={"providers": "ecb", "from": "2025-01-01", "to": "2025-12-31", "quotes": "idr"},
    timeout=30,
)
data = r.json()
print(r.status_code, len(data))
```

### Checking for missing data for these currencies 

```python
codes = "idr,usd,sgd,aud,jpy,gbp,cad,chf,krw,myr"

r = requests.get(
    f"{BASE}/rates",
    params={"providers": "ecb", "from": "2010-01-04", "to": "2010-01-08", "quotes": codes},
    timeout=30,
)
data = r.json()
print(r.status_code, len(data))
print(sorted({row["quote"] for row in data}))
print(min(row["date"] for row in data), max(row["date"] for row in data))
```
